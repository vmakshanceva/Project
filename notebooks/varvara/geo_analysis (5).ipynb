# Сбор студий звукозаписм

Я начала свою работу со сбора данных о студиях звукозаписи. Сначала я вручную выбрала 15 студий на сайте 2gis.ru и сделала таблицу из названия, адреса, рейтинга и номера телефона.

Следующим шагом я получила точные координаты этих студий с помощью кода. Координаты адресов, которые не дал мне код, я вбила вручную с помощью Yandex maps. Собрав все нужные данные, я сделала интерактивную карту с точками.

Данный код автоматически получает географические координаты для 15 студий звукозаписи в Москве. Он создаёт таблицу с названиями и адресами студий, затем для каждого адреса отправляет запрос к сервису Nominatim (OpenStreetMap), который ищет адрес и возвращает его координаты. Функция get_coordinates_osm формирует полный адрес с добавлением «Москва, Россия», отправляет запрос, обрабатывает ответ и извлекает координаты. Чтобы не перегружать сервис, между запросами делается пауза 1.5 секунды (time.sleep(1.5)). В конце код сохраняет полученную таблицу с координатами в файл studios_with_coords.csv и выводит результат на экран.

```python
import pandas as pd
import requests
import time
import re

# Функция для получения координат
def get_coordinates_osm(address):
    full_address = f"{address}, Москва, Россия"

    url = "https://nominatim.openstreetmap.org/search"
    params = {
        "q": full_address, # сервер ищет этот адрес
        "format": "json",
        "limit": 1 # вернуть только первый результат
    }
    headers = {
        "User-Agent": "MoscowStudioProject/1.0" # указывает имя программы, которая отправляет запрос
    }                                           # (Nominatim требует User-Agent, иначе может заблокировать)

    try: # блок, где могут возникнуть ошибки (если что-то пойдёт не так, программа перейдёт в except)
        response = requests.get(url, params=params, headers=headers)
        data = response.json() # превращает ответ сервера в словарь питона
        if data:
            lat = float(data[0]["lat"])
            lon = float(data[0]["lon"])
            print(f" {address[:40]}... → {lat:.4f}, {lon:.4f}")
            return lat, lon
        else:
            print(f" Не найдено: {address}")
            return None, None
    except Exception as e:
        print(f" Ошибка: {address}")
        return None, None

# Мои данные
data = {
    "name": [
        "Grusha Music", "Moon dragon", "KomorKing sound",
        "Продюсерский центр Игоря Сандлера", "Pioneer DJ School",
        "Хендрикс", "ЦАО Records", "Звук Во рту", "Вайб Рекордс",
        "333 sound", "Overtime", "Sandler Studio", "Major Studio",
        "All Music Studio", "Showbizrecords"
    ],
    "address": [
        "Большой Путинковский переулок, 5",
        "Петровский бульвар, 9 ст2",
        "Красный Октябрь, Болотная набережная, 3 ст2",
        "Потаповский переулок, 3 ст1",
        "Нижний Сусальный переулок, 5 ст4",
        "Марксистская улица, 34 к4",
        "Улица Земляной Вал, 50а ст3",
        "Улица Покровка, 2/1 ст2",
        "Нижняя Сыромятническая улица, 10 ст8",
        "Рождественский бульвар, 10/7 ст1",
        "Налесный переулок, 4",
        "Потаповский переулок, 3 ст1",
        "Нижняя Сыромятническая улица, 10 ст40",
        "Старокирочный переулок, 2",
        "Старая Басманная улица, 34"
    ]
}

df = pd.DataFrame(data) # превращает этот словарь в таблицу для удобства

# Получаем координаты
lat_list = []
lon_list = []

for address in df["address"]:
    lat, lon = get_coordinates_osm(address)
    lat_list.append(lat)
    lon_list.append(lon)
    time.sleep(1.5)

df["lat"] = lat_list
df["lon"] = lon_list

# Сохраняем
df.to_csv("studios_with_coords.csv", index=False)
print("\n Файл сохранён: studios_with_coords.csv")

print("\n Результат:")
print(df[["name", "address", "lat", "lon"]])
```

Тут я уже вручную исправляю координаты для тех адресов, которые Nominatim не смог найти. Он загружает ранее созданный файл studios_with_coords.csv, где у некоторых студий координаты остались пустыми, и заменяет их на значения, которые я уточнила через Яндекс.Карты. Для каждого адреса из словаря manual_coords код находит соответствующую строку в таблице и обновляет колонки lat (широта) и lon (долгота).

```python
import pandas as pd

df = pd.read_csv("studios_with_coords.csv")

# Уточнённые вручную координаты (с помощью яндекс карт)
manual_coords = {
    "Петровский бульвар, 9 ст2": (55.768588, 37.614779),
    "Красный Октябрь, Болотная набережная, 3 ст2": (55.740000, 37.609505),
    "Нижний Сусальный переулок, 5 ст4": (55.760206, 37.663889),
    "Улица Земляной Вал, 50а ст3": (55.751148, 37.655248),
    "Рождественский бульвар, 10/7 ст1": (55.766223, 37.626196),
    "Нижняя Сыромятническая улица, 10 ст40": (55.752794, 37.669782)
}

# Обновляем координаты в таблице
for address, (lat, lon) in manual_coords.items(): # штучка, чтобы в те места, где нет координат, вставились те, что я нашла вручную
    mask = df["address"] == address
    df.loc[mask, "lat"] = lat
    df.loc[mask, "lon"] = lon
    print(f"Обновлено: {address[:40]}... → {lat}, {lon}")

# Сохраняем обновлённый файл
df.to_csv("studios_with_coords.csv", index=False)
print("\n Файл studios_with_coords.csv обновлён")

# Проверяем, нет ли пустых
print(f"\n Пустых координат осталось: {df['lat'].isna().sum()}")

# Показываем результат
print("\n Итоговая таблица:")
print(df[["name", "address", "lat", "lon"]])
```

Медленно, но верно, мы подходим к ключевому этапу нашей работы. Код создаёт интерактивную карту Москвы с отмеченными на ней студиями звукозаписи. Он загружает файл studios_with_coords.csv, где уже есть все названия, адреса и координаты студий. Затем с помощью библиотеки folium создаёт карту с центром в центре Москвы и масштабом 13 (чтобы были видны районы). Для каждой студии из таблицы код добавляет на карту значок в точку с координатами lat и lon. При клике на значок всплывает окошко с названием студии и адресом. После добавления всех 15 точек код показывает карту в Colabе для удобства, а затем сохраняет карту в файл studios_map.html и скачивает его на компьютер.

```python
import pandas as pd
import folium
from google.colab import files
from IPython.display import display

# Загружаем файл
df = pd.read_csv("studios_with_coords.csv")

# Создаём карту
moscow_map = folium.Map(location=[55.7558, 37.6173], zoom_start=13) # команда, которую я использую для создания интерактивных карт

# Добавляем все точки
for _, row in df.iterrows(): # берёт таблицу df и превращает её в последовательность строк
    folium.Marker(
        location=[row["lat"], row["lon"]],
        popup=f"<b>{row['name']}</b><br>{row['address']}", # чтобы при клике на значок показывалось название и адрес
        icon=folium.Icon(color="red", icon="music", prefix="fa")
    ).add_to(moscow_map)

display(moscow_map)

# Сохраняем и скачиваем
moscow_map.save("studios_map.html")
files.download("studios_map.html")
```

Код загружает таблицу со студиями, а затем вручную задаётся список из пяти музыкальных локаций: два музыкальных вуза (Гнесинка и Консерватория), один концертный клуб («16 тонн») и два творческих кластера (Artplay и Винзавод).
```python
import folium
from IPython.display import display
from google.colab import files

df = pd.read_csv("studios_with_coords.csv")

# Музыкальные места
music_places = [
    ("РАМ им. Гнесиных", 55.7560, 37.5930, "вуз"),
    ("Московская консерватория", 55.7565, 37.6020, "вуз"),
    ("Клуб 16 тонн", 55.7610, 37.5730, "клуб"),
    ("Artplay", 55.7520, 37.6700, "кластер"),
    ("Винзавод", 55.7500, 37.6500, "кластер"),
]

# Цвета по типам места
color_by_type = {"вуз": "green", "клуб": "purple", "кластер": "blue"}

# Карта
moscow_map = folium.Map(location=[55.7558, 37.6173], zoom_start=13)

# Студии (они у нас красные)
for _, row in df.iterrows():
    folium.Marker(
        location=[row["lat"], row["lon"]],
        popup=f"<b>🎙️ {row['name']}</b><br>{row['address']}",
        icon=folium.Icon(color="red", icon="music", prefix="fa")
    ).add_to(moscow_map)

# Музыкальные места
for name, lat, lon, place_type in music_places:
    folium.Marker(
        location=[lat, lon],
        popup=f"<b>{name}</b><br>({place_type})",
        icon=folium.Icon(color=color_by_type.get(place_type, "gray"), icon="info-sign")
    ).add_to(moscow_map)

display(moscow_map)
moscow_map.save("studios_with_music.html")
files.download("studios_with_music.html")
```

# Тепловая карта

Данная карта нужна, чтобы определить плотность распределения уже существующих студий. Там где их много, появляется высокая конкурентность и открываться уже не выгодно. Там где их меньше — появляется потенциальный спот для открытия. Ниже показаны цвета и их обозначения на тепловой карте.

🔴 — много студий

🟡 — средняя плотность

🔵 — мало студий



Этот код создаёт тепловую карту плотности студий звукозаписи в Москве. Он загружает таблицу с координатами студий, которые я собрала ранее, и вместо отдельных маркеров показывает цветные пятна. Там, где студий много, образуется красное пятно, где мало — синее.

```python
import folium
from folium.plugins import HeatMap
from google.colab import files
import pandas as pd

# Загружаем данные
df = pd.read_csv("studios_with_coords.csv")

# Создаём карту
heat_map = folium.Map(location=[55.7558, 37.6173], zoom_start=12)

# Подготавливаем координаты для тепловой карты и берем только те строки, где координаты не пустые
heat_data = [[row["lat"], row["lon"]] for _, row in df.iterrows() if pd.notna(row["lat"])]

print(f" {len(heat_data)} точек добавлено в тепловую карту")

# Добавляем тепловую карту
HeatMap(
    heat_data,
    radius=15,        # радиус пятна (чем больше, тем шире размытие)
    blur=10,          # степень размытия
    min_opacity=0.5,  # минимальная прозрачность
    max_zoom=13       # масштаб
).add_to(heat_map)

# Показываем карту в колабе
display(heat_map)

heat_map.save("studios_heatmap.html")
files.download("studios_heatmap.html")
```

# Рейтинг районов

Подходя к концу моей части, мне нужно решить в каком же районе лучше открыть нашу студию. Для этого я решила воспользоваться следующей формулой: `Приоритет = Количество музыкальных мест / (Количество студий + 1)`.

Она показывает, насколько район перспективен для открытия новой студии. В числителе стоит количество музыкальных вузов, клубов, кластеров, где много музыкантов, а в знаменателе — количество существующих студий звукозаписи. Единица в знаменателе нужна, чтобы избежать деления на ноль, если в районе нет студий. Чем выше полученное значение, тем больше спрос превышает предложение, а значит, тем выгоднее открывать студию именно там. Если значение низкое, то конкурентов слишком много, и открываться не стоит.

```python
import pandas as pd
import matplotlib.pyplot as plt

# Загружаем студии
df_studios = pd.read_csv("studios_with_coords.csv")

# Список музыкальных мест из карты
music_places = [
    ("РАМ им. Гнесиных", 55.7560, 37.5930, "вуз"),
    ("Московская консерватория", 55.7565, 37.6020, "вуз"),
    ("Клуб 16 тонн", 55.7610, 37.5730, "клуб"),
    ("Artplay", 55.7520, 37.6700, "кластер"),
    ("Винзавод", 55.7500, 37.6500, "кластер"),
]

# Ручное распределение студий по районам по адресам из таблицы
studies_by_district = {
    "Басманный": 8,   # Потаповский, Покровка, Старокирочный, Сыромятническая и др.
    "Таганский": 3,   # Марксистская, Земляной Вал, Таганский
    "Тверской": 2,    # Большой Путинковский, Петровский бульвар
    "Мещанский": 1,   # Рождественский бульвар
    "Якиманка": 1,    # Болотная набережная
}

# Музыкальные места по районам
music_by_district = {
    "Басманный": 2,   # Artplay, Винзавод
    "Таганский": 0,
    "Тверской": 2,    # Гнесинка, Консерватория
    "Мещанский": 0,
    "Якиманка": 1,    # Клуб 16 тонн (ближе к Якиманке)
}

# табличка
df_rating = pd.DataFrame({
    "Район": list(studies_by_district.keys()), #Берёт ключи из словаря studies_by_district и превращает их в список
    "Количество студий": list(studies_by_district.values()),
    "Количество музыкальных мест": [music_by_district[r] for r in studies_by_district.keys()]
})

# Считаем приоритет
df_rating["Приоритет"] = df_rating["Количество музыкальных мест"] / (df_rating["Количество студий"] + 1)

# Сортируем по убыванию приоритета
df_rating = df_rating.sort_values("Приоритет", ascending=False) # по убыванию

# таблица вывод
print(df_rating.to_string(index=False))

# Строим график
plt.figure(figsize=(10, 6))
colors = ['red' if p < 0.3 else 'orange' if p < 0.6 else 'green' for p in df_rating["Приоритет"]]
plt.bar(df_rating["Район"], df_rating["Приоритет"], color=colors)
plt.xlabel("Район")
plt.ylabel("Приоритет (музыканты / студии)")
plt.title("Какой район лучше подходит для открытия студии")
plt.axhline(y=0.5, color='gray', linestyle='--', alpha=0.7)
for i, row in df_rating.iterrows():
    plt.text(row["Район"], row["Приоритет"] + 0.02, f"{row['Приоритет']:.2f}", ha='center')
plt.show()
```

# ООП



здесь приведены аналогичные карты, но с использованием ООП, поскольку того требует задание

```python
import pandas as pd
import folium
from folium.plugins import HeatMap
from IPython.display import display

class StudioGeographyAnalyzer:

    def __init__(self, csv_path="studios_with_coords.csv"): # загрузка данных из CSV-файла
        self.csv_path = csv_path
        self.df = pd.read_csv(csv_path)
        print(f"Загружено {len(self.df)} студий")

    def get_studios_count(self): # возвращает количество студий.
        return len(self.df)

    def get_studios_list(self):  # возвращает список студий, у которых есть координаты
        studios = []
        for _, row in self.df.iterrows():
            if pd.notna(row["lat"]) and pd.notna(row["lon"]):
                studios.append({
                    "name": row["name"],
                    "address": row["address"],
                    "lat": row["lat"],
                    "lon": row["lon"]
                })
        return studios

    def create_studios_map(self, zoom_start=13): # создаем карту со спотами студий
                                                 # zoom_start (int) - начальный масштаб карты

        m = folium.Map(location=[55.7558, 37.6173], zoom_start=zoom_start) # возвращает folium.Map (карта с красными отметками студий)

        for _, row in self.df.iterrows():
            if pd.notna(row["lat"]) and pd.notna(row["lon"]):
                folium.Marker(
                    location=[row["lat"], row["lon"]],
                    popup=f"<b>{row['name']}</b><br>{row['address']}",
                    icon=folium.Icon(color="red", icon="music", prefix="fa")
                ).add_to(m)

        return m

    def create_heatmap(self, radius=15, blur=10, zoom_start=12): # создаем тепловую карту плотности студий
        m = folium.Map(location=[55.7558, 37.6173], zoom_start=zoom_start)

        # готовим координаты
        heat_data = [[row["lat"], row["lon"]] for _, row in self.df.iterrows() if pd.notna(row["lat"])]

        HeatMap(
            heat_data,
            radius=radius,
            blur=blur,
            min_opacity=0.5
        ).add_to(m)

        return m

    def show_map(self, map_obj):
        display(map_obj)



geo = StudioGeographyAnalyzer("studios_with_coords.csv")

print("\n Карта со студиями:")
studios_map = geo.create_studios_map()
geo.show_map(studios_map)

print("\nТепловая карта плотности студий:")
heatmap = geo.create_heatmap()
geo.show_map(heatmap)

```

На основе этих данных я построила интерактивную карту Москвы с отметками всех студий, добавила на неё ключевые музыкальные места (вузы, клубы, творческие кластеры), а затем создала тепловую карту, которая показывает зоны высокой и низкой концентрации конкурентов. Дополнительно я разработала рейтинг районов по формуле приоритета, чтобы определить наилучшие районы для открытия новой студии. В результате анализ показал, что Тверской район (0.67) и Якиманка (0.50) имеют наилучшее соотношение спроса и предложения, в то время как Басманный район (0.22) перенасыщен студиями. На основе этих данных я считаю, что лучше всего открывать новую студию в Тверском районе, где высокая концентрация музыкантов и творческой молодёжи, но недостаточно студий.
